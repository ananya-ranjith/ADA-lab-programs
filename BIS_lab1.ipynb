```python

import random


population = [2, 5, 8, 3, 7]

for generation in range(5):


    fitness = population


    fitness.sort(reverse=True)
    parent1 = fitness[0]
    parent2 = fitness[1]


    child = (parent1 + parent2) // 2


    child = child + random.choice([-1, 0, 1])


    population[-1] = child

    print("Generation", generation + 1)
    print("Population:", population)
    print("Best:", max(population))
```

```python
import random


n = int(input("Enter number of bus stops: "))

print("Enter distance matrix:")
distance = []

for i in range(n):
    row = list(map(int, input().split()))
    distance.append(row)


def fitness(route):
    total = 0
    for i in range(n - 1):
        total += distance[route[i]][route[i + 1]]
    return total

population = []

for i in range(10):
    route = list(range(n))
    random.shuffle(route)
    population.append(route)


for generation in range(100):


    population.sort(key=fitness)
    parents = population[:5]

    new_population = parents.copy()

    while len(new_population) < 10:

        p1 = random.choice(parents)
        p2 = random.choice(parents)


        cut = random.randint(1, n - 1)
        child = p1[:cut]

        for stop in p2:
            if stop not in child:
                child.append(stop)


        if random.random() < 0.2:
            a, b = random.sample(range(n), 2)
            child[a], child[b] = child[b], child[a]

        new_population.append(child)

    population = new_population


best = min(population, key=fitness)

print("Best Bus Route:", [x + 1 for x in best])
print("Minimum Distance:", fitness(best), "km")
```
